```python
# Target variable
y = df['Attrition'].map({'No': 0, 'Yes': 1})

# Remove the target variable
X = df.drop('Attrition', axis=1)

print("Features:", X.shape)
print("Target:", y.shape)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Identify categorical columns
categorical_columns = X.select_dtypes(include=['object']).columns

# Identify numerical columns
numerical_columns = X.select_dtypes(exclude=['object']).columns

print("Categorical columns:", len(categorical_columns))
print("Numerical columns:", len(numerical_columns))
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_columns),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_columns)
    ]
)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

# Create the Logistic Regression pipeline
logistic_model = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('classifier', LogisticRegression(
            max_iter=1000,
            random_state=42
        ))
    ]
)

# Train the model
logistic_model.fit(X_train, y_train)

print("Logistic Regression model trained successfully!")
```

```python
# Predict employee attrition
y_pred = logistic_model.predict(X_test)

# Predict probability of attrition
y_pred_proba = logistic_model.predict_proba(X_test)[:, 1]

print("Predictions generated successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_pred_proba)

print(f"Accuracy : {accuracy:.3f}")
print(f"Precision: {precision:.3f}")
print(f"Recall   : {recall:.3f}")
print(f"F1 Score : {f1:.3f}")
print(f"ROC-AUC  : {roc_auc:.3f}")
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=['Stayed', 'Left']
)

disp.plot()
plt.title('Logistic Regression - Confusion Matrix')
plt.show()
```

```python
# Extract feature names after preprocessing

feature_names = logistic_model.named_steps['preprocessor'].get_feature_names_out()

# Extract Logistic Regression coefficients
coefficients = logistic_model.named_steps['classifier'].coef_[0]

# Create a dataframe
feature_importance = pd.DataFrame({
    'Feature': feature_names,
    'Coefficient': coefficients
})

# Add absolute coefficient
feature_importance['Absolute_Coefficient'] = (
    feature_importance['Coefficient'].abs()
)

# Sort by absolute coefficient
feature_importance = feature_importance.sort_values(
    'Absolute_Coefficient',
    ascending=False
)

feature_importance.head(20)
```

```python
# Select top positive and negative coefficients
top_positive = feature_importance.sort_values(
    'Coefficient', ascending=False
).head(10)

top_negative = feature_importance.sort_values(
    'Coefficient', ascending=True
).head(10)

# Combine them
top_features = pd.concat([
    top_positive,
    top_negative
]).sort_values('Coefficient')

# Clean feature names for visualization
top_features['Feature'] = (
    top_features['Feature']
    .str.replace('cat__', '', regex=False)
    .str.replace('num__', '', regex=False)
)

# Plot
plt.figure(figsize=(10, 8))

plt.barh(
    top_features['Feature'],
    top_features['Coefficient']
)

plt.axvline(x=0, linestyle='--')

plt.xlabel('Logistic Regression Coefficient')
plt.ylabel('Feature')
plt.title('Key Factors Associated with Employee Attrition')

plt.tight_layout()
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, confusion_matrix
)
```

```python

# Create Random Forest pipeline using the same preprocessor
rf_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ))
])
```

```python
# Train the model
rf_model.fit(X_train, y_train)

# Predictions
y_pred_rf = rf_model.predict(X_test)
y_proba_rf = rf_model.predict_proba(X_test)[:, 1]
```

```python
# Metrics
print("Random Forest — Threshold 0.50")
print("Accuracy :", round(accuracy_score(y_test, y_pred_rf), 3))
print("Precision:", round(precision_score(y_test, y_pred_rf), 3))
print("Recall   :", round(recall_score(y_test, y_pred_rf), 3))
print("F1 Score :", round(f1_score(y_test, y_pred_rf), 3))
print("ROC-AUC  :", round(roc_auc_score(y_test, y_proba_rf), 3))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_rf))
```

```python
# Random Forest with balanced class weights
rf_balanced = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        n_jobs=-1,
        class_weight='balanced'
    ))
])

rf_balanced.fit(X_train, y_train)

y_pred_rf_bal = rf_balanced.predict(X_test)
y_proba_rf_bal = rf_balanced.predict_proba(X_test)[:, 1]

print("Random Forest (balanced) — Threshold 0.50")
print("Accuracy :", round(accuracy_score(y_test, y_pred_rf_bal), 3))
print("Precision:", round(precision_score(y_test, y_pred_rf_bal), 3))
print("Recall   :", round(recall_score(y_test, y_pred_rf_bal), 3))
print("F1 Score :", round(f1_score(y_test, y_pred_rf_bal), 3))
print("ROC-AUC  :", round(roc_auc_score(y_test, y_proba_rf_bal), 3))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_rf_bal))
```

```python
# 1. Get predicted probabilities from the LR model
y_proba_lr = logistic_model.predict_proba(X_test)[:, 1]

# 2. Threshold sweep for balanced RF
thresholds = np.arange(0.10, 0.61, 0.05)
rf_threshold_results = []

for t in thresholds:
    y_pred_t = (y_proba_rf_bal >= t).astype(int)
    rf_threshold_results.append({
        'Threshold': round(t, 2),
        'Precision': round(precision_score(y_test, y_pred_t, zero_division=0), 3),
        'Recall':    round(recall_score(y_test, y_pred_t, zero_division=0), 3),
        'F1':        round(f1_score(y_test, y_pred_t, zero_division=0), 3)
    })

rf_threshold_df = pd.DataFrame(rf_threshold_results)
print("Random Forest (balanced) — Threshold Sweep")
print(rf_threshold_df.to_string(index=False))

# 3. ROC curves for LR and RF (balanced)
from sklearn.metrics import roc_curve
fpr_lr, tpr_lr, _ = roc_curve(y_test, y_proba_lr)
fpr_rf, tpr_rf, _ = roc_curve(y_test, y_proba_rf_bal)

auc_lr = roc_auc_score(y_test, y_proba_lr)
auc_rf = roc_auc_score(y_test, y_proba_rf_bal)

plt.figure(figsize=(7, 6))
plt.plot(fpr_lr, tpr_lr, label=f'Logistic Regression (AUC = {auc_lr:.3f})', color='#1f77b4')
plt.plot(fpr_rf, tpr_rf, label=f'Random Forest balanced (AUC = {auc_rf:.3f})', color='#d62728')
plt.plot([0, 1], [0, 1], linestyle='--', color='grey', label='Random classifier')

plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve Comparison — Logistic Regression vs Random Forest')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
from sklearn.ensemble import HistGradientBoostingClassifier

# HistGradientBoosting pipeline
hgb_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', HistGradientBoostingClassifier(
        random_state=42,
        max_iter=200,
        learning_rate=0.05,
        class_weight='balanced'  # available in scikit-learn >= 1.2
    ))
])

hgb_model.fit(X_train, y_train)

y_pred_hgb = hgb_model.predict(X_test)
y_proba_hgb = hgb_model.predict_proba(X_test)[:, 1]

print("HistGradientBoosting — Threshold 0.50")
print("Accuracy :", round(accuracy_score(y_test, y_pred_hgb), 3))
print("Precision:", round(precision_score(y_test, y_pred_hgb), 3))
print("Recall   :", round(recall_score(y_test, y_pred_hgb), 3))
print("F1 Score :", round(f1_score(y_test, y_pred_hgb), 3))
print("ROC-AUC  :", round(roc_auc_score(y_test, y_proba_hgb), 3))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_hgb))

# Quick threshold sweep for HGB
import numpy as np
thresholds = np.arange(0.10, 0.61, 0.05)
hgb_results = []
for t in thresholds:
    y_pred_t = (y_proba_hgb >= t).astype(int)
    hgb_results.append({
        'Threshold': round(t, 2),
        'Precision': round(precision_score(y_test, y_pred_t, zero_division=0), 3),
        'Recall':    round(recall_score(y_test, y_pred_t, zero_division=0), 3),
        'F1':        round(f1_score(y_test, y_pred_t, zero_division=0), 3)
    })

import pandas as pd
hgb_df = pd.DataFrame(hgb_results)
print("\nHistGradientBoosting — Threshold Sweep")
print(hgb_df.to_string(index=False))
```
