# 📘 Reference Paper Overview for Presentation

**Title:** Forecasting Foreign Exchange Rate with Machine Learning Techniques: Chinese Yuan to US Dollar Using XGBoost and LSTM Model
**Authors:** Usman Ullah, Zhensheng Huang, Dawood Rehman, Sulaiman Khan, Haroon Rashid, Imran Ullah (2024)

## 🎯 1. Research Objectives & Core Questions
**Primary Goal:** To investigate how effectively modern machine learning and sequential deep learning models can forecast future values of the highly volatile Chinese Yuan (CNY) to US Dollar (USD) exchange rate.

**Core Research Question:** How do non-linear machine learning ensembles (**XGBoost**) compare against recurrent neural networks (**LSTM**) in handling the non-stationarity, extreme fluctuations, and historical sequence dependencies unique to foreign currency markets?

**Economic Impact:** Providing reliable tools for central bank policymakers (monetary framework control), international trade businesses (hedging currency risk), and global market investors.

---

## 🛠️ 2. Methodology & Experimental Framework

### 📊 A. Data Scope & Preprocessing
**The Dataset:** Covers a 36-month timeline of daily historical exchange rate data stretching from **April 25, 2020, to May 26, 2023**.
**Outlier Cleansing:** Implemented a **Z-Score filter** ($|z| > 3$) to separate artificial data entry errors from true historical structural shocks (like economic crises), smoothly interpolation-averaging baseline noise.
**Stationarity Transformation:** Conducted an **Augmented Dickey-Fuller (ADF) Test**. Because the raw daily trend returned a non-stationary profile ($p > 0.05$), **first-differencing** was applied to yield stationary conditions ($p = 0.0$, ADF stat: $-28.83$).
**Scaling:** Handled data bounds using **Min-Max Scaling** to map target vectors strictly onto a $[0, 1]$ interval, guaranteeing accelerated gradient descent optimization and fast training convergence.
**Data Partitioning:** Chronologically separated the processed sequence into an **80% Training split** and a **20% Testing split**.

### 💻 B. Models Selected
1. **Long Short-Term Memory (LSTM):** Selected for its gated recurrent cell structure, which mitigates vanishing gradients and implicitly captures long-range sequential contexts and structural path dependencies across irregular time windows.
2. **eXtreme Gradient Boosting (XGBoost):** Selected as a regularized tree-boosting benchmark that minimizes structural overfitting while prioritizing features automatically on highly multi-dimensional tabular matrix inputs.

---

## 📈 3. Empirical Results & Performance Evaluation

The models were evaluated against actual test market data across four key regression errors. The exact results from **Table 2** of the paper are outlined below:

| Metric | Deep Learning: LSTM | Machine Learning: XGBoost |
| :--- | :---: | :---: |
| **MAE** (Mean Absolute Error) | **0.043** | 0.111 |
| **MSE** (Mean Squared Error) | **0.003** | 0.701 |
| **RMSE** (Root Mean Squared Error) | **0.055** | 1.119 |
| **MAPE** (Mean Absolute Percentage Error) | 0.595% | **0.330%** |

### 🔍 Core Findings & Conclusions
**LSTM Superiority:** The paper concludes that **LSTM is the superior framework** for univariable time-series forecasting, yielding significantly lower absolute error traces (RMSE of $0.055$ vs. XGBoost's $1.119$).
**The Sequential Factor:** While XGBoost prevents overfitting and manages immediate structural data parameters well, it lacks native recurrent hidden cell blocks to infer continuous, underlying historical temporal properties hidden inside continuous sequences.
**Baseline Benchmarks:** Both models significantly outperformed traditional linear tools like ARIMA and standard Moving Averages, as basic models lack the capacity to capture non-linear market shocks and long-term correlations simultaneously.

This cell imports necessary Python libraries for data manipulation, web scraping financial data, and sets up global configuration variables like start and end dates for data fetching, along with definitions for various macroeconomic indicators from FRED and World Bank. It also includes a patch for a pandas deprecation warning to ensure compatibility.

```python
import pandas.util._decorators as _pd_dec

if not callable(getattr(_pd_dec, "deprecate_kwarg", None)):
    def _noop_deprecate_kwarg(old_arg_name, new_arg_name,
                              mapping=None, stacklevel=2):
        """No-op shim replacing the removed pandas helper."""
        def decorator(func):
            return func
        return decorator
    _pd_dec.deprecate_kwarg = _noop_deprecate_kwarg

# ─────────────────────────────────────────────────────────────────
# 1.  STANDARD IMPORTS
# ─────────────────────────────────────────────────────────────────
import warnings
import pandas as pd
import pandas_datareader.data as web   # requires: pip install pandas_datareader
import pandas_datareader.wb as wb_reader # Use pandas_datareader for World Bank data

warnings.filterwarnings("ignore", category=FutureWarning)


# ─────────────────────────────────────────────────────────────────
# 2.  CONFIGURATION
# ─────────────────────────────────────────────────────────────────
START_DATE = "2008-01-01"
END_DATE    = "2025-12-31"

# FRED Monthly indicator codes and their human-readable column names
FRED_MONTHLY_INDICATORS: dict[str, str] = {
    "CPIAUCSL"         : "US_Inflation_CPI_pct",   # US CPI (monthly)
    "CPALTT01CNM659N"  : "CN_Inflation_CPI_pct",   # China CPI (monthly)
    "M2SL"             : "US_M2_pctGDP",           # US M2 (monthly)
}
FRED_MONTHLY_CHINA_M2_SERIES = "M2CH" # China M2 from FRED (monthly, but often fails)

# World Bank Annual indicator codes and their human-readable column names
# Exclude indicators we try to get monthly from FRED (CPI, M2) and daily (CNY/USD)
WB_ANNUAL_INDICATORS: dict[str, str] = {
    "NY.GDP.MKTP.KD.ZG"    : "GDP_Growth_pct",           # GDP growth, annual %
    "FR.INR.RINR"          : "RealInterestRate_pct",     # Real interest rate, %
    "BN.CAB.XOKA.GD.ZS"    : "CurrentAccount_pctGDP",   # Current account balance, % GDP
    "GC.DOD.TOTL.GD.ZS"    : "GovDebt_pctGDP",          # General government debt, % GDP
    "BX.TRF.PWKR.DT.GD.ZS": "Remittances_pctGDP",      # Personal remittances received, % GDP
}

COUNTRIES = ["CN", "US"]   # ISO-2 country codes


# ─────────────────────────────────────────────────────────────────
# 3.  STEP 1 — FETCH DAILY CNY/USD FROM FRED
#     DEXCHUS: China / U.S. Foreign Exchange Rate (yuan per dollar)
#     Source: https://fred.stlouisfed.org/series/DEXCHUS
# ─────────────────────────────────────────────────────────────────
print("=" * 60)
print("Step 1: Fetching CNY/USD daily rate from FRED …")
print("=" * 60)

fred_df: pd.DataFrame = web.DataReader(
    "DEXCHUS",       # FRED series ID
    "fred",          # data source
    START_DATE,
    END_DATE,
)

# Rename for clarity; drop any weekends/holidays that FRED returns as NaN
fred_df.columns = ["CNYUSD"]
fred_df.index   = pd.to_datetime(fred_df.index)
fred_df.index.name = "Date"
fred_df = fred_df.dropna(subset=["CNYUSD"])   # remove non-trading days

print(f"  FRED rows fetched : {len(fred_df):,}")
print(f"  Date range        : {fred_df.index.min().date()} → {fred_df.index.max().date()}")
print(f"  Sample values:\n{fred_df.head(3)}\n")


# ─────────────────────────────────────────────────────────────────
# 4.  STEP 2 — FETCH MONTHLY MACRO INDICATORS FROM FRED
# ─────────────────────────────────────────────────────────────────
print("=" * 60)
print("Step 2: Fetching monthly macro indicators from FRED …")
print("=" * 60)

fred_monthly_dfs = []

# Fetch US & China CPI, US M2
for fred_series_id, col_name in FRED_MONTHLY_INDICATORS.items():
    print(f"  Fetching {col_name} ({fred_series_id})...")
    try:
        data = web.DataReader(fred_series_id, "fred", START_DATE, END_DATE)
        data.columns = [col_name]
        fred_monthly_dfs.append(data)
    except Exception as e:
        print(f"  Error fetching {fred_series_id} from FRED: {e}")

# Fetch China M2 with World Bank fallback
print(f"  Fetching China M2 ({FRED_MONTHLY_CHINA_M2_SERIES}) or from World Bank as fallback...")
china_m2_df = pd.DataFrame()
try:
    china_m2_fred = web.DataReader(FRED_MONTHLY_CHINA_M2_SERIES, "fred", START_DATE, END_DATE)
    china_m2_fred.columns = ["CN_M2_pctGDP"]
    china_m2_df = china_m2_fred
    print("  China M2 data fetched successfully from FRED.")
except Exception as e:
    print(f"  Warning: Could not fetch China M2 from FRED ({FRED_MONTHLY_CHINA_M2_SERIES}). Error: {e}")
    print("  Attempting to fetch China M2 from World Bank (annual as monthly data is elusive)...")
    try:
        # Indicator for Broad money (M2) in current LCU for China
        wb_m2_raw = wb_reader.download(indicator='FM.LBL.BMNY.CN', country=['CN'],
                             start=int(START_DATE[:4]), end=int(END_DATE[:4]))
        # Process into a DataFrame with 'Date' index and 'CN_M2_pctGDP' column
        wb_m2_raw = wb_m2_raw.reset_index()
        wb_m2_raw['year'] = pd.to_datetime(wb_m2_raw['year'], format='%Y')
        wb_m2_processed = wb_m2_raw.set_index('year')['FM.LBL.BMNY.CN'].to_frame(name="CN_M2_pctGDP")
        wb_m2_processed.index.name = 'Date'
        china_m2_df = wb_m2_processed
        print("  China M2 data fetched successfully from World Bank (annual data).")
    except Exception as e_wb:
        print(f"  Warning: Could not fetch China M2 from World Bank. Error: {e_wb}")
        print("  China M2 data will be missing.")

if not china_m2_df.empty:
    fred_monthly_dfs.append(china_m2_df)

fred_monthly_df = pd.concat(fred_monthly_dfs, axis=1)
fred_monthly_df.index.name = "Date"
# For monthly data, ffill can fill gaps between reported months for consolidation purposes later
fred_monthly_df = fred_monthly_df.ffill()

print(f"  FRED monthly data rows fetched: {len(fred_monthly_df):,}")
if not fred_monthly_df.empty:
    print(f"  Date range        : {fred_monthly_df.index.min().date()} → {fred_monthly_df.index.max().date()}")
    print(f"  Sample values:\n{fred_monthly_df.head(3)}\n")


# ─────────────────────────────────────────────────────────────────
# 5.  STEP 3 — FETCH ANNUAL MACRO INDICATORS FROM WORLD BANK (remaining indicators)
#     Using pandas_datareader.wb.download
#     Fetching each indicator separately to avoid URL length/timeout issues.
# ─────────────────────────────────────────────────────────────────
print("=" * 60)
print("Step 3: Fetching annual macro indicators from World Bank (remaining indicators) …")
print("=" * 60)

wb_raw_list_annual = []
for indicator_code, indicator_name in WB_ANNUAL_INDICATORS.items():
    print(f"  Fetching {indicator_name} ({indicator_code})...")
    try:
        data = wb_reader.download(
            indicator=indicator_code, # Fetch one indicator at a time
            country=COUNTRIES,
            start=int(START_DATE[:4]),
            end=int(END_DATE[:4]),
        )
        # The returned DataFrame has a MultiIndex (country, year) and one column (indicator_code).
        # We need to flatten this so 'country' and 'year' become regular columns.
        data = data.reset_index()
        # Rename 'country' to 'economy' and 'year' to 'time'
        data = data.rename(columns={'country': 'economy', 'year': 'time'})
        # Rename the actual indicator column to 'value' and add an 'indicator_code' column
        data = data.rename(columns={indicator_code: 'value'}) # Rename the actual indicator column to 'value'
        data['indicator_code'] = indicator_code # Add a column to identify the indicator
        wb_raw_list_annual.append(data)
    except Exception as e:
        print(f"  Error fetching {indicator_code} from World Bank: {e}")

# Concatenate all fetched dataframes and then pivot
if wb_raw_list_annual:
    wb_raw_concat_annual = pd.concat(wb_raw_list_annual, ignore_index=True)
    # Now pivot to get indicators as columns
    wb_raw_annual = wb_raw_concat_annual.pivot_table(index=['economy', 'time'], columns='indicator_code', values='value')
    wb_raw_annual = wb_raw_annual.reset_index()
else:
    wb_raw_annual = pd.DataFrame() # Create an empty DataFrame if no data was fetched

# Process wb_raw_annual into wb_annual_processed:
wb_annual_processed = pd.DataFrame() # Initialize as empty
if not wb_raw_annual.empty:
    # Convert 'time' to datetime objects for proper time-series handling
    wb_raw_annual['time'] = pd.to_datetime(wb_raw_annual['time'], format='%Y')

    # Rename indicator columns using the WB_ANNUAL_INDICATORS map
    # Only include indicators that are actually present in wb_raw_annual
    indicator_rename_map_annual = {k:v for k,v in WB_ANNUAL_INDICATORS.items() if k in wb_raw_annual.columns}
    wb_raw_annual = wb_raw_annual.rename(columns=indicator_rename_map_annual)

    # Pivot to get country-specific columns for each indicator
    # e.g., 'CN_GDP_Growth_pct', 'US_GDP_Growth_pct'
    wb_annual_processed = wb_raw_annual.pivot_table(index='time', columns='economy', values=list(indicator_rename_map_annual.values()))

    # Flatten the MultiIndex columns (indicator_name, country_code) to (country_code_indicator_name)
    wb_annual_processed.columns = [f"{col[1]}_{col[0]}" for col in wb_annual_processed.columns]
    wb_annual_processed.columns.name = None # Remove the column MultiIndex name

    # Ensure the index name is 'Date' to align with fred_df for merging
    wb_annual_processed.index.name = "Date"

print(f"  World Bank annual data rows fetched: {len(wb_annual_processed):,}")
if not wb_annual_processed.empty:
    print(f"  Date range        : {wb_annual_processed.index.min().date()} → {wb_annual_processed.index.max().date()}")
    print(f"  Sample values:\n{wb_annual_processed.head(3)}\n")


# ─────────────────────────────────────────────────────────────────
# 6.  STEP 4 — CONSOLIDATE ALL DATA TO DAILY FREQUENCY AND MERGE
# ─────────────────────────────────────────────────────────────────
print("=" * 60)
print("Step 4: Consolidating all data to daily frequency and merging …")
print("=" * 60)

# Create a daily date range covering the entire period
full_date_range = pd.date_range(start=START_DATE, end=END_DATE, freq='D')

# Prepare list of dataframes to concatenate/merge, all reindexed to full_date_range and ffilled
all_macro_data_sources_daily = []

# FRED daily exchange rate is already daily
all_macro_data_sources_daily.append(fred_df)

# Reindex monthly FRED data to daily and forward-fill
if not fred_monthly_df.empty:
    fred_monthly_daily = fred_monthly_df.reindex(full_date_range, method='ffill')
    all_macro_data_sources_daily.append(fred_monthly_daily)

# Reindex annual World Bank data to daily and forward-fill
if not wb_annual_processed.empty:
    wb_annual_daily = wb_annual_processed.reindex(full_date_range, method='ffill')
    all_macro_data_sources_daily.append(wb_annual_daily)

# Merge all datasets into final_df
final_df = pd.concat(all_macro_data_sources_daily, axis=1)

# Dropped lines for `final_df.dropna(inplace=True)` and associated print statement

print(f"  Final consolidated DataFrame rows: {len(final_df):,}")
print(f"  Date range        : {final_df.index.min().date()} → {final_df.index.max().date()}")
print(f"  Sample values:\n{final_df.head(3)}\n")
print(f"  Missing values per column:\n{final_df.isnull().sum()}")

# Save the combined dataframe
output_path = 'cnyusd_macro_features.csv'
final_df.to_csv(output_path, index=True)
print(f"Combined data saved to {output_path}")

print("\nConsolidated data pipeline executed successfully with enhanced frequency.")

# ─────────────────────────────────────────────────────────────────
# 7.  STEP 5 — (OPTIONAL) M2 DATA FROM WORLD BANK IF FRED FAILS
#     This is handled in Step 2 for China's M2 data.
# ─────────────────────────────────────────────────────────────────

print("\n" + "=" * 60)
print("Verification: US_Inflation_CPI_pct")
print("=" * 60)

# Displaying raw monthly data for US_Inflation_CPI_pct from fred_monthly_df
if 'US_Inflation_CPI_pct' in fred_monthly_df.columns:
    print("\n--- Original FRED Monthly Data (US_Inflation_CPI_pct) ---")
    print(fred_monthly_df['US_Inflation_CPI_pct'].head())
    print(fred_monthly_df['US_Inflation_CPI_pct'].tail())
    print(fred_monthly_df['US_Inflation_CPI_pct'].describe())
else:
    print("US_Inflation_CPI_pct not found in fred_monthly_df.")

# Displaying final daily data for US_Inflation_CPI_pct from final_df
if 'US_Inflation_CPI_pct' in final_df.columns:
    print("\n--- Final Consolidated Daily Data (US_Inflation_CPI_pct) ---")
    print(final_df['US_Inflation_CPI_pct'].head())
    print(final_df['US_Inflation_CPI_pct'].tail())
    print(final_df['US_Inflation_CPI_pct'].describe())
else:
    print("US_Inflation_CPI_pct not found in final_df.")

print("\nBased on FRED series 'CPIAUCSL', this represents the Consumer Price Index for All Urban Consumers: All Items in U.S. City Average. The data is fetched monthly and then forward-filled to daily frequency in the final DataFrame. The values and frequency appear consistent with this definition.")
```

### Fetching Real Interest Rates
This cell defines a function `fetch_real_interest_rates` which is designed to retrieve nominal interest rates and inflation data for both the US and China from various sources (FRED and World Bank). It then calculates the real interest rate for each country by subtracting inflation from the nominal rate. The function handles data processing and consolidation to provide a clean time series of real interest rates.

```python
import pandas as pd
import pandas_datareader.data as web
import pandas_datareader.wb as wb
import datetime

def fetch_real_interest_rates(country, start_date_str, end_date_str):
    """
    Fetches nominal rates and inflation from FRED and World Bank,
    and calculates the real interest rate.
    """
    rate_data = {}

    start_date_dt = pd.to_datetime(start_date_str)
    end_date_dt = pd.to_datetime(end_date_str)

    if country.upper() == 'US':
        try:
            # Fetch nominal rate
            fed_funds = web.DataReader('FEDFUNDS', 'fred', start_date_dt, end_date_dt)
            rate_data['nominal_rate_df'] = fed_funds # Store as df
            rate_data['nominal_rate_col'] = 'FEDFUNDS' # Store original column name

            # Fetch inflation (CPI)
            cpi = web.DataReader('CPIAUCSL', 'fred', start_date_dt, end_date_dt)
            # Calculate monthly inflation rate from CPI
            # CPIAUCSL is monthly, so pct_change() will give monthly inflation
            inflation = cpi.pct_change() * 100
            rate_data['inflation_df'] = inflation
            rate_data['inflation_col'] = 'CPIAUCSL' # Original column name
        except Exception as e:
            print(f"Error fetching US data: {e}")
            return None

    elif country.upper() == 'CN':
        try:
            # Fetch nominal rate from FRED. Fallback to World Bank if necessary.
            discount_rate = web.DataReader('INTDSRCNM193N', 'fred', start_date_dt, end_date_dt)
            rate_data['nominal_rate_df'] = discount_rate
            rate_data['nominal_rate_col'] = 'INTDSRCNM193N'

            # Fetch inflation from World Bank (annual)
            # wb.download expects integer years
            inflation_df_raw = wb.download(indicator='FP.CPI.TOTL.ZG', country=['CN'], start=int(start_date_str[:4]), end=int(end_date_str[:4]))

            # Process World Bank data to a time-indexed DataFrame
            inflation_df_raw = inflation_df_raw.reset_index()
            inflation_df_raw['year'] = pd.to_datetime(inflation_df_raw['year'], format='%Y')
            inflation_series = inflation_df_raw.set_index('year')['FP.CPI.TOTL.ZG']

            # Convert to monthly and forward fill.
            # We need to create a date range to resample to monthly
            monthly_index = pd.date_range(start=start_date_str, end=end_date_str, freq='MS')
            inflation_processed_series = inflation_series.reindex(monthly_index, method='ffill')

            # Store as DataFrame with a consistent column name for 'inflation'
            rate_data['inflation_df'] = inflation_processed_series.to_frame(name='inflation')
            rate_data['inflation_col'] = 'inflation' # The new column name in the processed DF
        except Exception as e:
            print(f"Error fetching China data: {e}")
            return None

    # Check if required data is present
    if 'nominal_rate_df' not in rate_data or rate_data['nominal_rate_df'] is None or \
       'inflation_df' not in rate_data or rate_data['inflation_df'] is None:
        print(f"Cannot calculate real rate for {country}: missing nominal rate or inflation data.")
        return None

    # Rename columns in individual dataframes before concatenation
    nominal_df = rate_data['nominal_rate_df'].rename(columns={rate_data['nominal_rate_col']: 'nominal'})
    inflation_df = rate_data['inflation_df'].rename(columns={rate_data['inflation_col']: 'inflation'})

    # Align indexes and concatenate
    combined = pd.concat([nominal_df, inflation_df], axis=1)

    # Ensure both columns have data before attempting arithmetic
    if combined['nominal'].isnull().all() or combined['inflation'].isnull().all():
        print(f"Cannot calculate real rate for {country}: one of the series is entirely NaN after concatenation.")
        return None

    combined['real_rate'] = combined['nominal'] - combined['inflation']
    combined.fillna(method='ffill', inplace=True)
    combined.dropna(inplace=True)

    if combined.empty:
        print(f"No valid real rate data after processing for {country}.")
        return None

    return combined[['real_rate']]

# --- Usage for your project ---
start = '2008-01-01'
end = '2023-12-31'

us_real_rate = fetch_real_interest_rates('US', start, end)
china_real_rate = fetch_real_interest_rates('CN', start, end)

if us_real_rate is not None:
    print(us_real_rate.head())
else:
    print("US real rate data could not be fetched or processed.")

if china_real_rate is not None:
    print(china_real_rate.head())
else:
    print("China real rate data could not be fetched or processed.")
```

### Handling Missing Data
This step focuses on addressing missing values in the dataset. For the high-frequency `CNYUSD` exchange rate, linear interpolation is used to fill gaps, preserving its temporal trend. For macroeconomic indicators, forward-fill followed by backward-fill ensures that all nulls are handled, assuming that the last observed value is the best estimate for subsequent missing periods.

```python
# ── 2.1  HANDLING MISSING DATA ───────────────────────────────────────────────
section("STEP 1 — HANDLING MISSING DATA  (§3.4.I)")

# Exchange rate → linear interpolation (preserves trend in high-freq data)
pre_null = df[TARGET].isna().sum()
df[TARGET] = df[TARGET].interpolate(method="linear", limit_direction="both")
post_null = df[TARGET].isna().sum()
log(f"CNYUSD  — nulls before: {pre_null}, after linear interp: {post_null}")

# Macroeconomic indicators → forward-fill captures temporal consistency.
# True seasonal-decomposition imputation is unnecessary here because the
# annual WB values were already forward-filled in the data collection step
# and have zero nulls; we guard anyway.
for c in MACRO_COLS:
    n_null = df[c].isna().sum()
    if n_null > 0:
        df[c] = df[c].ffill().bfill()
        log(f"{c:45s}  — {n_null} nulls filled with ffill/bfill", indent=1)
    else:
        log(f"{c:45s}  — 0 nulls  ✓", indent=1)
```

### Outlier Detection and Treatment
This cell implements a two-pronged approach for outlier detection: the Interquartile Range (IQR) method and Z-score analysis. For the `CNYUSD` series, potential outliers are flagged but not removed, as they might represent significant regime shifts. For macroeconomic features, extreme Z-scores (absolute value > 4) are treated by replacing them with smoothed values using a Savitzky-Golay filter, preventing erroneous data from distorting the models.

```python
# ── 2.2  OUTLIER DETECTION & TREATMENT ──────────────────────────────────────
section("STEP 2 — OUTLIER DETECTION & TREATMENT  (§3.4.II)")
log("Method: IQR (1.5× fence) AND z-score (|z|>3) on CNYUSD daily series.")
log("Macro cols have annual step-changes (14 unique values each) that are")
log("not errors — they are retained as valid structural features.")

# CNYUSD — the only true high-frequency series
Q1, Q3 = df[TARGET].quantile([0.25, 0.75])
IQR     = Q3 - Q1
z_scores = np.abs(sp_stats.zscore(df[TARGET]))

iqr_mask = (df[TARGET] < Q1 - 1.5 * IQR) | (df[TARGET] > Q3 + 1.5 * IQR)
z_mask   = z_scores > 3
combo    = iqr_mask | z_mask

n_iqr = iqr_mask.sum()
n_z   = z_mask.sum()
n_both = combo.sum()

log(f"CNYUSD outliers — IQR: {n_iqr} | z>3: {n_z} | union: {n_both}", indent=1)

if n_both > 0:
    # Cross-check: flag but retain — these likely correspond to regime events
    log("  → Flagging as 'outlier_flag' column; no values replaced.", indent=1)
    log("    (Cross-check dates below against known CNY policy events)", indent=1)
    df["CNYUSD_outlier_flag"] = combo.astype(int)
    log(f"  Outlier dates: {list(df.index[combo].date)}", indent=2)
else:
    log("  → No outliers detected in CNYUSD series.  ✓", indent=1)
    df["CNYUSD_outlier_flag"] = 0

# Macro cols — replace ONLY erroneous extreme z-scores (|z|>4, not structural)
log()
log("Macro columns — checking for erroneous (|z|>4) data errors:")
for c in MACRO_COLS:
    z = np.abs(sp_stats.zscore(df[c].dropna()))
    extreme = (z > 4).sum()
    if extreme > 0:
        # LOESS-style smoothing with Savitzky-Golay (window=11, poly=2)
        smoothed = savgol_filter(df[c].fillna(method="ffill"), window_length=11,
                                  polyorder=2)
        mask = np.abs(sp_stats.zscore(df[c].fillna(method="ffill"))) > 4
        df.loc[mask, c] = smoothed[mask]
        log(f"  {c}: {extreme} extreme points replaced with SG smoothing", indent=1)
    else:
        log(f"  {c}: no extreme errors  ✓", indent=1)
```

### Logarithmic Transformation
This step applies a logarithmic transformation to the `CNYUSD` target variable, creating `CNYUSD_log`. This transformation can help stabilize variance and make the data more normally distributed. It also calculates the log returns (`CNYUSD_logret`), which approximate daily percentage changes and are often more stationary, a desirable property for time series modeling.

```python
# ── 3.1  LOGARITHMIC TRANSFORMATION ─────────────────────────────────────────
section("STEP 3 — LOGARITHMIC TRANSFORMATION  (§3.7.I)")

df["CNYUSD_log"]    = np.log(df[TARGET])
df["CNYUSD_logret"] = df["CNYUSD_log"].diff()   # ≈ daily % change

log(f"CNYUSD_log    : log-level series created")
log(f"CNYUSD_logret : log-return (diff of log-level) created")
log(f"  logret skew     = {df['CNYUSD_logret'].skew():.4f}")
log(f"  logret kurtosis = {df['CNYUSD_logret'].kurtosis():.4f}")
```

### Stationarity Tests and Differencing
This critical step addresses the stationarity of the time series data. The Augmented Dickey-Fuller (ADF) test is performed on `CNYUSD` and all macroeconomic columns to check for unit roots. If a series is found to be non-stationary (p-value > 0.05), it is first-differenced (`.diff()`) to achieve stationarity, which is a common requirement for many time series models to ensure reliable predictions.

```python
# ── 3.2  ADF STATIONARITY TEST + DIFFERENCING ────────────────────────────────
section("STEP 4 — STATIONARITY TESTS & DIFFERENCING  (§3.7.II)")
log("ADF test (H0: unit root present; p<0.05 → reject → stationary)")
log(f"{'Column':<45}  {'p-raw':>8}  {'status':>12}  {'p-diff':>8}")
log("-" * 80)

DIFF_COLS: list[str] = []      # columns that needed differencing

adf_summary: dict[str, dict] = {}

# Test the raw CNYUSD level
p_cny = adfuller(df[TARGET].dropna(), autolag="AIC")[1]
status = "STATIONARY" if p_cny <= 0.05 else "UNIT ROOT"
log(f"  {TARGET:<43}  {p_cny:>8.4f}  {status:>12}")
adf_summary[TARGET] = {"p_raw": p_cny, "differenced": False}

# If non-stationary, first-difference it
if p_cny > 0.05:
    df["CNYUSD_diff"] = df[TARGET].diff()
    p2 = adfuller(df["CNYUSD_diff"].dropna(), autolag="AIC")[1]
    log(f"    → After 1st diff: p={p2:.4f}  ({'STATIONARY' if p2<=0.05 else 'STILL NS'})")
    adf_summary[TARGET]["p_diff"] = p2
    adf_summary[TARGET]["differenced"] = True

# Test log return (should already be stationary)
p_lr = adfuller(df["CNYUSD_logret"].dropna(), autolag="AIC")[1]
log(f"  {'CNYUSD_logret':<43}  {p_lr:>8.4f}  {'STATIONARY' if p_lr<=0.05 else 'UNIT ROOT':>12}")

# Test each macro column
for c in MACRO_COLS:
    ser = df[c].dropna()

    # Skip ADF test if the series is constant (all values are the same)
    if ser.nunique() <= 1:
        p = 0.0 # Consider constant series as stationary
        stat = "STATIONARY"
        p_diff_str = "—"
        log(f"  {c:<43}  {p:>8.4f}  {stat:>12}  {p_diff_str:>8}  (Constant series)")
    else:
        p   = adfuller(ser, autolag="AIC")[1]
        stat = "STATIONARY" if p <= 0.05 else "UNIT ROOT"
        p_diff_str = "—"

        if p > 0.05:
            diff_ser = ser.diff().dropna()
            # Also check if differenced series is constant before testing
            if diff_ser.nunique() <= 1:
                p2 = 0.0 # Consider constant differenced series as stationary
                log(f"    → After 1st diff: p={p2:.4f}  (STATIONARY - constant after diff)")
            else:
                p2 = adfuller(diff_ser, autolag="AIC")[1]
                log(f"    → After 1st diff: p={p2:.4f}  ({'STATIONARY' if p2<=0.05 else 'STILL NS'})")
            p_diff_str = f"{p2:.4f}"
            df[c + "_diff"] = df[c].diff()
            DIFF_COLS.append(c)

        log(f"  {c:<43}  {p:>8.4f}  {stat:>12}  {p_diff_str:>8}")
    adf_summary[c] = {"p_raw": p, "differenced": c in DIFF_COLS}

log()
log(f"Columns differenced: {len(DIFF_COLS)}")
for c in DIFF_COLS:
    log(f"  {c}_diff created", indent=1)
```

### Standardization and Normalization
This cell prepares numerical features for machine learning models by applying scaling techniques. The `CNYUSD` target variable is Min-Max normalized to a range of 0-1, which is often beneficial for neural networks like LSTMs. All macroeconomic columns are Z-score standardized, transforming them to have a mean of 0 and a standard deviation of 1, which helps various algorithms perform better by treating all features equally.

```python
# ── 3.3  STANDARDIZATION & NORMALIZATION ─────────────────────────────────────
section("STEP 5 — STANDARDIZATION & NORMALIZATION  (§3.4.III)")
log("CNYUSD        → Min-Max normalization  (range 0–1)")
log("Macro columns → Z-score standardization  (mean=0, std=1)")

mm_scaler = MinMaxScaler()
df["CNYUSD_norm"] = mm_scaler.fit_transform(df[[TARGET]])
log(f"  CNYUSD_norm range: [{df['CNYUSD_norm'].min():.4f}, {df['CNYUSD_norm'].max():.4f}]")

z_scaler  = StandardScaler()
z_cols    = [c + "_z" for c in MACRO_COLS]
df[z_cols] = z_scaler.fit_transform(df[MACRO_COLS])
log(f"  Z-scored columns : {len(z_cols)}")
```

### Creating Lagged Variables and Rolling Statistics
This step engineers new features based on past values of the target variable and macroeconomic differentials. Various lagged versions of `CNYUSD` and its log returns are created (e.g., `CNYUSD_lag1`, `CNYUSD_logret_lag5`). Rolling means and standard deviations are also calculated to capture momentum and volatility. Additionally, lagged differentials between US and China macroeconomic indicators are generated to provide comparative insights, enriching the feature set for forecasting models.

```python
# ── 3.4  LAGGED VARIABLES ─────────────────────────────────────────────────────
section("STEP 6 — LAGGED VARIABLES  (§3.7.III)")
log("Lag lengths explored: 1, 2, 3, 5, 10, 21 trading days")
log("AIC-optimal lag selected via AR(p) model on CNYUSD_logret")

# AIC-based optimal lag selection (AR model up to 21 lags)
from statsmodels.tsa.ar_model import AutoReg

logret_clean = df["CNYUSD_logret"].dropna()
aic_scores   = {}
for p in range(1, 22):
    try:
        m = AutoReg(logret_clean, lags=p, old_names=False).fit()
        aic_scores[p] = m.aic
    except Exception:
        pass

optimal_lag = min(aic_scores, key=aic_scores.get)
log(f"  AIC-optimal lag = {optimal_lag}  (AIC = {aic_scores[optimal_lag]:.4f})")

LAG_SET = sorted(set([1, 2, 3, 5, 10, 21, optimal_lag]))
log(f"  Creating lag set: {LAG_SET}")

for lag in LAG_SET:
    df[f"CNYUSD_lag{lag}"]        = df[TARGET].shift(lag)
    df[f"CNYUSD_logret_lag{lag}"] = df["CNYUSD_logret"].shift(lag)

# Rolling statistics (capture momentum / volatility)
for w in [5, 10, 21]:
    df[f"CNYUSD_roll{w}_mean"] = df[TARGET].rolling(w).mean()
    df[f"CNYUSD_roll{w}_std"]  = df[TARGET].rolling(w).std()

log(f"  Lag columns created : {2 * len(LAG_SET)}")
log(f"  Rolling stat columns: 6  (mean + std × windows 5/10/21)")

# Lagged macro differentials
diff_pairs = [
    ("CN_Inflation_CPI_pct",          "US_Inflation_CPI_pct",                "DIFF_Inflation"),
    ("CN_M2_pctGDP",                  "US_M2_pctGDP",                        "DIFF_M2"),
    ("China_GDP_Growth_pct",          "United States_GDP_Growth_pct",        "DIFF_GDP"),
    ("China_RealInterestRate_pct",    "United States_RealInterestRate_pct",  "DIFF_RealIR"),
    ("China_CurrentAccount_pctGDP",   "United States_CurrentAccount_pctGDP", "DIFF_CA"),
    ("China_Remittances_pctGDP",      "United States_Remittances_pctGDP",    "DIFF_Remit"),
]
for cn_c, us_c, name in diff_pairs:
    if cn_c in df.columns and us_c in df.columns:
        df[name] = df[cn_c] - df[us_c]
        for lag in [1, 5, 21]:
            df[f"{name}_lag{lag}"] = df[name].shift(lag)

log(f"  Macro differentials + lagged versions created")


# Drop rows with NaN introduced by lags/rolling
df_model = df.dropna()
log()
log(f"After dropping NaN rows: {df_model.shape[0]} rows remain  "
    f"(lost {len(df)-len(df_model)} rows from lag/rolling)")
```

### Correlation Analysis for Feature Selection
This cell performs an initial feature selection step using correlation analysis. It calculates the absolute correlation between each candidate feature (lagged variables, Z-scored macro features, differentials) and the `CNYUSD` target. Features with a very low absolute correlation (less than 0.01) are identified and dropped, as they are unlikely to contribute significantly to the model's predictive power. This helps in reducing noise and model complexity.

```python
# ─────────────────────────────────────────────────────────────────────────────
# 4.  SECTION 3.5 — FEATURE SELECTION
# ─────────────────────────────────────────────────────────────────────────────

# ── 4.1  CORRELATION ANALYSIS ─────────────────────────────────────────────────
section("STEP 7 — CORRELATION ANALYSIS  (§3.5.I)")

# Columns to include: lag features, macro z-scores, differentials
CANDIDATE_FEATURES = (
    [f"CNYUSD_lag{l}"        for l in LAG_SET] +
    [f"CNYUSD_logret_lag{l}" for l in LAG_SET] +
    [f"CNYUSD_roll{w}_mean" for w in [5, 10, 21]] +
    [f"CNYUSD_roll{w}_std"  for w in [5, 10, 21]] +
    z_cols +
    [p[2] for p in diff_pairs] +
    [f"{p[2]}_lag{l}" for p in diff_pairs for l in [1, 5, 21]]
)
# Keep only those that exist in df_model
CANDIDATE_FEATURES = [f for f in CANDIDATE_FEATURES if f in df_model.columns]

corr_with_target = (
    df_model[CANDIDATE_FEATURES + [TARGET]]
    .corr()[TARGET]
    .drop(TARGET)
    .abs()
    .sort_values(ascending=False)
)

log(f"Correlations with {TARGET} (top 20):")
log(f"{'Feature':<40}  {'|r|':>6}")
log("-" * 50)
for feat, r in corr_with_target.head(20).items():
    log(f"  {feat:<38}  {r:>6.4f}")

# Drop features with |r| < 0.01 (near-zero predictive power)
LOW_CORR_THRESHOLD = 0.01
drop_low_corr = corr_with_target[corr_with_target < LOW_CORR_THRESHOLD].index.tolist()
log()
log(f"Features dropped (|r| < {LOW_CORR_THRESHOLD}): {len(drop_low_corr)}")
for c in drop_low_corr:
    log(f"  {c}", indent=1)

SELECTED_FEATURES = [f for f in CANDIDATE_FEATURES if f not in drop_low_corr]
log(f"Features after correlation filter: {len(SELECTED_FEATURES)}")
```

### XGBoost Feature Importance for Selection
Building on the correlation analysis, this step uses an XGBoost Regressor to assess feature importance. A temporary train/validation split is created to train a basic XGBoost model, and its feature importances (F-scores) are calculated. Features are then selected based on a cumulative importance threshold (e.g., retaining features that account for 99% of the total importance), or a minimum/maximum number of features to ensure a robust set for modeling.

```python
# ── 4.2  XGBOOST FEATURE IMPORTANCE ──────────────────────────────────────────
section("STEP 8 — XGBOOST FEATURE IMPORTANCE  (§3.5.II)")

# Temporary split just to train the importance model
_n = len(df_model)
_t = int(_n * 0.70)
_v = int(_n * 0.85)

X_tr = df_model.iloc[:_t][SELECTED_FEATURES]
y_tr = df_model.iloc[:_t][TARGET]
X_vl = df_model.iloc[_t:_v][SELECTED_FEATURES]
y_vl = df_model.iloc[_t:_v][TARGET]

xgb_selector = xgb.XGBRegressor(
    n_estimators  = 300,
    max_depth     = 4,
    learning_rate = 0.05,
    subsample     = 0.8,
    colsample_bytree = 0.8,
    random_state  = 42,
    verbosity     = 0,
)
xgb_selector.fit(
    X_tr, y_tr,
    eval_set=[(X_vl, y_vl)],
    verbose=False,
)

fi = (
    pd.Series(xgb_selector.feature_importances_, index=SELECTED_FEATURES)
    .sort_values(ascending=False)
)

log(f"Top-15 XGBoost feature importances (F-score, gain-based):")
log(f"{'Feature':<40}  {'Importance':>10}")
log("-" * 54)
for feat, score in fi.head(15).items():
    bar = "█" * int(score * 100)
    log(f"  {feat:<38}  {score:>10.4f}  {bar}")

# Retain features with cumulative importance ≥ 99% (or top-20, whichever is smaller)
fi_cumsum    = fi.cumsum()
cum_threshold = 0.99
keep_fi = fi[fi_cumsum <= cum_threshold].index.tolist()
if len(keep_fi) < 5:                         # ensure minimum 5 features
    keep_fi = fi.head(5).index.tolist()
if len(keep_fi) > 25:                        # cap at 25
    keep_fi = fi.head(25).index.tolist()

log()
log(f"Features retained (cumulative importance ≥ {cum_threshold*100:.0f}%): {len(keep_fi)}")
SELECTED_FEATURES = keep_fi
```

### PCA for Dimensionality Reduction on Macro Features
This cell applies Principal Component Analysis (PCA) specifically to the Z-scored macroeconomic features. PCA is used here to reduce the dimensionality of the macro features while retaining most of their variance. Principal components that collectively explain a specified percentage of the variance (e.g., 90%) are selected and added to the feature set. This helps to mitigate multicollinearity and reduce the number of input features for the models.

```python
# ── 4.3  PCA — DIMENSIONALITY REDUCTION ──────────────────────────────────────
section("STEP 9 — PCA ON MACRO FEATURES  (§3.5.III)")
log("PCA applied to z-scored macro columns only.")
log("Principal components explaining 90% of macro variance are retained.")

MACRO_Z_COLS = [c for c in z_cols if c in df_model.columns]
pca = PCA(n_components=0.90, random_state=42)
pca_arr = pca.fit_transform(df_model[MACRO_Z_COLS])

PCA_COL_NAMES = [f"PC{i+1}" for i in range(pca.n_components_)]
df_pca = pd.DataFrame(pca_arr, index=df_model.index, columns=PCA_COL_NAMES)
df_model = pd.concat([df_model, df_pca], axis=1)

log(f"  Input macro features  : {len(MACRO_Z_COLS)}")
log(f"  PCA components kept   : {pca.n_components_}")
log(f"  Variance explained    : {pca.explained_variance_ratio_.sum()*100:.2f}%")
log()
log(f"  {'Component':<10}  {'Var. explained':>16}  {'Cumulative':>12}")
log("  " + "-" * 44)
cum = 0
for i, ev in enumerate(pca.explained_variance_ratio_):
    cum += ev
    log(f"  {'PC'+str(i+1):<10}  {ev*100:>14.2f}%  {cum*100:>10.2f}%")

# Add PCA columns to selected features
SELECTED_FEATURES = SELECTED_FEATURES + PCA_COL_NAMES
```

### XGBoost Modeling
This section focuses on developing and evaluating an XGBoost (Extreme Gradient Boosting) model for time series forecasting. It prepares the data by splitting it into features (X) and target (y) for training, validation, and testing. A `GridSearchCV` with `TimeSeriesSplit` is used for hyperparameter tuning to find the optimal XGBoost configuration. The best model's performance is then evaluated on the test set using various metrics (MAE, RMSE, R-squared, MAPE), and its forecasts are visualized against the actual values.

```python
from sklearn.model_selection import GridSearchCV, TimeSeriesSplit
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import xgboost as xgb
import matplotlib.pyplot as plt
import numpy as np

section("STEP 14 — XGBOOST MODELING")
log("Preparing data for XGBoost...")

# Access the split datasets
train_data = df_final[df_final.split == 'train']
val_data = df_final[df_final.split == 'validation']
test_data = df_final[df_final.split == 'test']

X_train, y_train = train_data[FINAL_FEATURE_COLS], train_data[TARGET]
X_val, y_val = val_data[FINAL_FEATURE_COLS], val_data[TARGET]
X_test, y_test = test_data[FINAL_FEATURE_COLS], test_data[TARGET]

log(f"  Training features shape: {X_train.shape}, target shape: {y_train.shape}")
log(f"  Validation features shape: {X_val.shape}, target shape: {y_val.shape}")
log(f"  Test features shape: {X_test.shape}, target shape: {y_test.shape}")

log("\nPerforming Hyperparameter Tuning for XGBoost using GridSearchCV...")

# Combine train and validation for tuning, as is common before final test evaluation
# This also ensures GridSearchCV has a larger dataset for CV splits.
X_train_val = pd.concat([X_train, X_val])
y_train_val = pd.concat([y_train, y_val])

# Define the parameter grid for GridSearchCV
param_grid = {
    'n_estimators': [100, 200, 300], # Reduced for faster execution
    'learning_rate': [0.05, 0.1],
    'max_depth': [3, 5],
    'subsample': [0.8],
    'colsample_bytree': [0.8]
}

# Initialize XGBoost Regressor
xgbr = xgb.XGBRegressor(objective='reg:squarederror', random_state=42, n_jobs=-1)

# Use TimeSeriesSplit for cross-validation given the temporal nature of the data
tscv = TimeSeriesSplit(n_splits=5) # Example: 5 splits

grid_search = GridSearchCV(estimator=xgbr, param_grid=param_grid,
                           cv=tscv, scoring='neg_mean_squared_error',
                           verbose=1, n_jobs=-1)

grid_search.fit(X_train_val, y_train_val)

log(f"  Best parameters found: {grid_search.best_params_}")
log(f"  Best RMSE on validation sets: {np.sqrt(-grid_search.best_score_):.4f}")

# Get the best model
best_xgboost_model = grid_search.best_estimator_

log("\nEvaluating Best XGBoost Model on Test Set...")

# Make predictions on the test set
test_predictions_xgb = best_xgboost_model.predict(X_test)

# Evaluate test predictions
mae_test_xgb = mean_absolute_error(y_test, test_predictions_xgb)
rmse_test_xgb = np.sqrt(mean_squared_error(y_test, test_predictions_xgb))
r2_test_xgb = r2_score(y_test, test_predictions_xgb)

# Function to calculate MAPE (Mean Absolute Percentage Error)
def mean_absolute_percentage_error(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    # Avoid division by zero
    return np.mean(np.abs((y_true - y_pred) / y_true)) * 100

mape_test_xgb = mean_absolute_percentage_error(y_test, test_predictions_xgb)

log(f"  Test MAE (XGBoost): {mae_test_xgb:.4f}")
log(f"  Test RMSE (XGBoost): {rmse_test_xgb:.4f}")
log(f"  Test R-squared (XGBoost): {r2_test_xgb:.4f}")
log(f"  Test MAPE (XGBoost): {mape_test_xgb:.4f}%")

log("\nVisualizing XGBoost Forecasts vs Actuals...")

# Create a DataFrame for test predictions to align with original indices
test_predictions_xgb_series = pd.Series(test_predictions_xgb, index=y_test.index)

plt.figure(figsize=(18, 9))
plt.plot(y_train.index, y_train, label='Train Actual', color='blue', alpha=0.7)
plt.plot(y_val.index, y_val, label='Validation Actual', color='orange', alpha=0.7)
plt.plot(y_test.index, y_test, label='Test Actual', color='green', alpha=0.7)
plt.plot(test_predictions_xgb_series.index, test_predictions_xgb_series,
         label='Test Forecast (XGBoost)', color='red', linestyle='--', alpha=0.7)
plt.title('XGBoost Forecast vs Actuals for CNYUSD')
plt.xlabel('Date')
plt.ylabel('CNY/USD')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

### LSTM Modeling
This section details the development and evaluation of an LSTM (Long Short-Term Memory) neural network model for time series forecasting. It involves preparing the data by scaling the target variable and creating sequences of past observations, which is essential for LSTM inputs. A grid search is performed to tune LSTM hyperparameters such as `look_back` (number of time steps), `lstm_units`, `batch_size`, and `epochs`. The best model is then evaluated on the test set using MAE, RMSE, R-squared, and MAPE, and its predictions are visualized.

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.optimizers import Adam
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import itertools

section("STEP 15 — LSTM MODELING")
log("Preparing data for LSTM...")

# Access the split datasets from df_final
train_data = df_final[df_final.split == 'train']
val_data = df_final[df_final.split == 'validation']
test_data = df_final[df_final.split == 'test']

# Separate features (X) and target (y)
X_train_pre = train_data[FINAL_FEATURE_COLS]
y_train_pre = train_data[TARGET]
X_val_pre = val_data[FINAL_FEATURE_COLS]
y_val_pre = val_data[TARGET]
X_test_pre = test_data[FINAL_FEATURE_COLS]
y_test_pre = test_data[TARGET]

log(f"  Training features shape: {X_train_pre.shape}, target shape: {y_train_pre.shape}")
log(f"  Validation features shape: {X_val_pre.shape}, target shape: {y_val_pre.shape}")
log(f"  Test features shape: {X_test_pre.shape}, target shape: {y_test_pre.shape}")

# Scale the target variable separately, as it's not scaled in FINAL_FEATURE_COLS
y_scaler = MinMaxScaler(feature_range=(0, 1))
y_train_scaled = y_scaler.fit_transform(y_train_pre.values.reshape(-1, 1))
y_val_scaled = y_scaler.transform(y_val_pre.values.reshape(-1, 1))
y_test_scaled = y_scaler.transform(y_test_pre.values.reshape(-1, 1))

# Features are already scaled (z-scored or min-max normalized) in FINAL_FEATURE_COLS
# Ensure they are numpy arrays
X_train = X_train_pre.values
X_val = X_val_pre.values
X_test = X_test_pre.values

# Function to create time series sequences for LSTM
def create_sequences(X, y, look_back):
    Xs, ys = [], []
    for i in range(len(X) - look_back):
        Xs.append(X[i:(i + look_back)])
        ys.append(y[i + look_back])
    return np.array(Xs), np.array(ys)

log("  Creating sequences for LSTM...")

# Hyperparameters for tuning
param_grid = {
    'look_back': [5, 10, 15],  # Number of past time steps to consider
    'lstm_units': [50, 100],  # Number of LSTM units
    'batch_size': [32, 64],
    'epochs': [20, 30]
}

best_score = float('inf')
best_params = {}
best_model = None

# Perform a simple grid search
log("  Performing hyperparameter tuning for LSTM...")
for look_back, lstm_units, batch_size, epochs in itertools.product(
    param_grid['look_back'], param_grid['lstm_units'],
    param_grid['batch_size'], param_grid['epochs']
): # Removed param_grid['learning_rate'] as requested
    # Using a fixed learning rate since it's excluded from tuning
    learning_rate = 0.01
    log(f"    Trying params: look_back={look_back}, lstm_units={lstm_units}, batch_size={batch_size}, epochs={epochs}, lr={learning_rate})")

    # Create sequences for current look_back
    X_train_seq, y_train_seq = create_sequences(X_train, y_train_scaled, look_back)
    X_val_seq, y_val_seq = create_sequences(X_val, y_val_scaled, look_back)

    # Reshape input to be [samples, time steps, features]
    n_features = X_train.shape[1]
    X_train_seq = X_train_seq.reshape(X_train_seq.shape[0], look_back, n_features)
    X_val_seq = X_val_seq.reshape(X_val_seq.shape[0], look_back, n_features)

    # Build LSTM model
    model = Sequential()
    model.add(LSTM(units=lstm_units, activation='relu', input_shape=(look_back, n_features)))
    model.add(Dropout(0.2))
    model.add(Dense(units=1))
    model.compile(optimizer=Adam(learning_rate=learning_rate), loss='mse')

    # Train the model
    history = model.fit(X_train_seq, y_train_seq, epochs=epochs, batch_size=batch_size,
                        validation_data=(X_val_seq, y_val_seq), verbose=0, shuffle=False)

    # Evaluate on validation set
    val_loss = model.evaluate(X_val_seq, y_val_seq, verbose=0)
    val_rmse = np.sqrt(val_loss)

    log(f"      Validation RMSE: {val_rmse:.4f}")

    if val_rmse < best_score:
        best_score = val_rmse
        best_params = {'look_back': look_back, 'lstm_units': lstm_units,
                       'batch_size': batch_size, 'epochs': epochs, 'learning_rate': learning_rate}
        best_model = model

log(f"  Best LSTM parameters found: {best_params}")
log(f"  Best Validation RMSE: {best_score:.4f}")

log("\nEvaluating Best LSTM Model on Test Set...")

# Retrain model with best parameters on combined train+val data if necessary,
# but for simplicity, we'll use the best model found during validation tuning.
# In a production setting, a final retraining on train+val before testing is good practice.

# Create test sequences using the best look_back
X_test_seq, y_test_seq = create_sequences(X_test, y_test_scaled, best_params['look_back'])
X_test_seq = X_test_seq.reshape(X_test_seq.shape[0], best_params['look_back'], n_features)

# Make predictions
test_predictions_scaled = best_model.predict(X_test_seq)

# Inverse transform predictions to original scale
test_predictions_lstm = y_scaler.inverse_transform(test_predictions_scaled).flatten()

# Align actuals for evaluation (adjust for look_back offset)
y_test_aligned_lstm = y_test_pre.iloc[best_params['look_back']:]

# Ensure predictions and actuals have the same length and index
min_len = min(len(y_test_aligned_lstm), len(test_predictions_lstm))
y_test_aligned_lstm = y_test_aligned_lstm.iloc[:min_len]
test_predictions_lstm = test_predictions_lstm[:min_len]

# Evaluate test predictions
mae_test_lstm = mean_absolute_error(y_test_aligned_lstm, test_predictions_lstm)
rmse_test_lstm = np.sqrt(mean_squared_error(y_test_aligned_lstm, test_predictions_lstm))
r2_test_lstm = r2_score(y_test_aligned_lstm, test_predictions_lstm)

def mean_absolute_percentage_error(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    # Avoid division by zero by replacing 0 with a small epsilon
    return np.mean(np.abs((y_true - y_pred) / np.maximum(y_true, 1e-8))) * 100

mape_test_lstm = mean_absolute_percentage_error(y_test_aligned_lstm, test_predictions_lstm)

log(f"  Test MAE (LSTM): {mae_test_lstm:.4f}")
log(f"  Test RMSE (LSTM): {rmse_test_lstm:.4f}")
log(f"  Test R-squared (LSTM): {r2_test_lstm:.4f}")
log(f"  Test MAPE (LSTM): {mape_test_lstm:.4f}%")

log("\nVisualizing LSTM Forecasts vs Actuals...")

# Create a DataFrame for test predictions to align with original indices
test_predictions_lstm_series = pd.Series(test_predictions_lstm, index=y_test_aligned_lstm.index)

plt.figure(figsize=(18, 9))
plt.plot(y_train_pre.index, y_train_pre, label='Train Actual', color='blue', alpha=0.7)
plt.plot(y_val_pre.index, y_val_pre, label='Validation Actual', color='orange', alpha=0.7)
plt.plot(y_test_aligned_lstm.index, y_test_aligned_lstm, label='Test Actual', color='green', alpha=0.7)
plt.plot(test_predictions_lstm_series.index, test_predictions_lstm_series,
         label=f'Test Forecast (LSTM {best_params["look_back"]} timesteps)', color='purple', linestyle='--', alpha=0.7)
plt.title(f'LSTM Forecast vs Actuals for CNYUSD (Best Lookback: {best_params["look_back"]})')
plt.xlabel('Date')
plt.ylabel('CNY/USD')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
